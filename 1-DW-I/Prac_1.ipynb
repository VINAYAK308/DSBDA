```python
# convert a categorical column (lunch) into numerical form using One-Hot Encoding and then add it back to the original DataFrame.
# pd.get_dummies() performs One-Hot Encoding.
# It converts each unique category in the lunch column into a separate binary (0/1) column.
```

```python
df_Lunch = pd.get_dummies(df['lunch'])
df_new = pd.concat([df, df_Lunch], axis=1)#pd.concat() joins DataFrames together.axis=1 means join column-wise.
#This adds the new one-hot encoded columns to the original DataFrame.
df_new #for print 
```
