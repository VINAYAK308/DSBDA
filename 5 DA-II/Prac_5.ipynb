# Description of below code
: → selects all rows

[2, 3] → selects column index 2 and column index 3

.values → converts the selected data into a NumPy array

```python
x = dataset.iloc[:, [2, 3]].values
y = dataset.iloc[:, 4].values
```

```python
print(x)
```

# This function is used to split the dataset into training and testing sets.

It helps evaluate how well a Machine Learning model performs on unseen data.                                                   x → independent variables / features
y → dependent variable / target
These two are split together, so each x sample correctly matches its y label.                                                           
random_state = 0
Controls the randomness of the split
Ensures reproducibility
Using the same random_state always gives the same train-test split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.30, random_state = 0)
```

# Feature Scaling?
Many ML algorithms (e.g., Logistic Regression, SVM, KNN) are sensitive to the scale of features
If one feature has large values and another has small values, the model may become biased
Standardization puts all features on the same scale                                                                               
sc = StandardScaler()
Creates a StandardScaler object
This object will learn the mean and standard deviation from the data                                                           

x_train = sc.fit_transform(x_train)
fit → calculates mean & standard deviation from training data
transform → scales the training data using those values
Both steps combined using fit_transform, Scaling is learned only from training data to avoid data leakage                        
This code standardizes features so that all variables have mean 0 and standard deviation 1, improving model performance and fairness.

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

# classifier = LogisticRegression(random_state = 0)

Creates a Logistic Regression model object
random_state = 0 ensures reproducible results
The model will learn how input features relate to output classes                                                            

                                                                    

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
classifier.fit(x_train, y_train)
```

#
.predict() uses the trained Logistic Regression model

It predicts the class labels for the given input data    

x_test contains unseen (test) feature data
This data was not used during training
It helps check how well the model generalizes                                                                                     
y_pred stores the predicted output values
These are the model’s guesses based on learned patterns        

```python
y_pred = classifier.predict(x_test)
```

# print(y_pred)
    
Displays the predicted class labels generated by the model
These are the outputs predicted from x_test

```python
print(y_pred)
```

# print(y_test)

Displays the actual (true) class labels
These are the correct answers from the dataset

```python
print(y_test)
```

#
confusion_matrix(y_test, y_pred)
Takes:
y_test → actual (true) labels
y_pred → predicted labels
Returns a 2×2 matrix for binary classification                                                                                 

Where:
TN (True Negative) → Correctly predicted 0
FP (False Positive) → Incorrectly predicted 1
FN (False Negative) → Incorrectly predicted 0
TP (True Positive) → Correctly predicted 1

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

# This means:

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

# Precision measures how many of the positive predictions made by the model are actually correct.

Formula:

Precision= TP+FP/TP

Where:
TP (True Positive) → Correctly predicted positive cases
FP (False Positive) → Incorrectly predicted positive cases                                                                     85.71 % -Correctness of positive predictions (positive predictions made by the model are actually correct.)          

```python
from sklearn.metrics import precision_score, recall_score
precision_score(y_test, y_pred)
```

# Recall measures how many of the actual positive cases were correctly identified by the model.                                                                                                          
Formula:

Recall=TP+FN/TP

Where:
TP (True Positive) → Correctly predicted positives
FN (False Negative) → Positives predicted as negative                                                                            FN (False Negative) → Positives predicted as negative                                                                              

```python
recall_score(y_test, y_pred)
```

# f1_score(y_test, y_pred) do?

Compares:
y_test → actual labels
y_pred → predicted labels
Computes the harmonic mean of precision and recall
By default, calculates F1-score for positive class (1)

```python
from sklearn.metrics import f1_score
f1_score(y_test, y_pred)
```

# Summary Table
Metric	What it measures

F1-score	Balance of precision & recall
